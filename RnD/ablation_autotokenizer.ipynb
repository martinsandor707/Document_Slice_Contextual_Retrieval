```python
import os
import json
import hashlib
from pathlib import Path
from typing import List, Dict, Any
from transformers import AutoTokenizer
from docling.document_converter import DocumentConverter
import transformers

transformers.logging.set_verbosity_error()

# TODO: Set your folder path containing PDF files
PDF_FOLDER = "input"  # TODO: FILL THIS IN
OUTPUT_FOLDER = "split_documents/AutoTokenizer"
EMBEDDING_MODEL_NAME = "nomic-ai/nomic-embed-text-v1.5" # TODO: FILL THIS IN

CHUNK_SIZE = 2000 
CHUNK_OVERLAP = 200

converter = DocumentConverter()

def chunk_text(text: str, tokenizer:AutoTokenizer, chunk_size: int, chunk_overlap: int) -> List[str]:
	"""Chunk text using Huggingface's AutoTokenizer."""
	# Tokenize the entire text
	tokens = tokenizer.encode(text, add_special_tokens=False)
	
	chunks = []
	start = 0
	
	while start < len(tokens):
		end = start + chunk_size
		chunk_tokens = tokens[start:end]
		chunk_text = tokenizer.decode(chunk_tokens)
		chunks.append(chunk_text)
		start += chunk_size - chunk_overlap
	
	return chunks

def generate_sha256(text: str) -> str:
	"""Generate SHA256 hash for a text chunk."""
	return hashlib.sha256(text.encode('utf-8')).hexdigest()

def process_pdfs(folder_path: str, output_folder: str = "output_chunks"):
	"""Scan all PDF files in a folder and process them."""
	
	# Initialize tokenizer
	tokenizer = AutoTokenizer.from_pretrained(EMBEDDING_MODEL_NAME)
	
	# Create output folder if it doesn't exist
	os.makedirs(output_folder, exist_ok=True)
	
	# Get all PDF files in the folder
	pdf_folder = Path(folder_path)
	pdf_files = list(pdf_folder.glob("*.pdf"))
	
	print(f"Found {len(pdf_files)} PDF files")
	
	for pdf_file in pdf_files:
		print(f"Processing: {pdf_file.name}")
		
		# Step 1: Convert PDF to markdown (without chunking)
		markdown_text = converter.convert(str(pdf_file)).document.export_to_markdown(image_placeholder="")
		
		# Step 2: Chunk the markdown text using AutoTokenizer
		chunks = chunk_text(markdown_text, tokenizer, CHUNK_SIZE, CHUNK_OVERLAP)
		
		# Step 3: Create JSON objects for each chunk
		chunk_data = []
		for chunk in chunks:
			chunk_obj = {
				"text": chunk,
				"document": pdf_file.name,
				"id": generate_sha256(chunk)
			}
			chunk_data.append(chunk_obj)
		
		# Step 4: Save to JSON file (one per document)
		output_filename = pdf_file.stem + ".json"
		output_path = os.path.join(output_folder, output_filename)
		
		with open(output_path, 'w', encoding='utf-8') as f:
			json.dump(chunk_data, f, indent=2, ensure_ascii=False)
		
		print(f"  Saved {len(chunks)} chunks to {output_path}")
	
	print("\nProcessing complete!")
```

```python
process_pdfs(PDF_FOLDER, OUTPUT_FOLDER)
```
